```python
from google.cloud import bigquery
from google.cloud import bigquery_storage

bq_client = bigquery.Client(project="chotot-dwh")
bqs_client = bigquery_storage.BigQueryReadClient()
pty_ad_query = '''
select *
from temp.huy__pty_ad
'''
ad_performance_query = '''
select ad_id, date, adview_count
from chotot_data.ad_daily
where date >= '2025-05-01'
'''
order_service_query = '''
select *
from temp.Huy__order_service
'''
pty_ad_df = bq_client.query(pty_ad_query).to_arrow(bqstorage_client=bqs_client).to_pandas()
ad_performance_df = bq_client.query(ad_performance_query).to_arrow(bqstorage_client=bqs_client).to_pandas()
order_service_df = bq_client.query(order_service_query).to_arrow(bqstorage_client=bqs_client).to_pandas()
```

```python
import pandas as pd
no_ads_by_account_id = pty_ad_df.groupby("account_id").agg(no_ads=("ad_id", "count")).reset_index()
pty_ad_df = pd.merge(pty_ad_df, no_ads_by_account_id, on="account_id", how="left")
```

```python
import sqlite3

def write_dataframe_to_sqlite(df: pd.DataFrame, db_path: str, table_name: str):
    """
    Write a Pandas DataFrame to an SQLite database.

    Args:
        df (pd.DataFrame): The DataFrame to write.
        db_path (str): Path to the SQLite database file.
        table_name (str): Name of the table to write data into.
    """
    with sqlite3.connect(db_path) as conn:
        df.to_sql(table_name, conn, if_exists='replace', index=False)
        print(f"Data written to table '{table_name}' in database '{db_path}'.")

ad_info = pd.read("ad_info.csv")
ad_performance = pd.read("ad_performance.csv")
order_service = pd.read("order_service.csv")

write_dataframe_to_sqlite(ad_info.fillna(0), "hackathon.db", "ad_info")
write_dataframe_to_sqlite(ad_performance.fillna(0), "hackathon.db", "ad_performance")
write_dataframe_to_sqlite(order_service.fillna(0), "hackathon.db", "orders")

```
